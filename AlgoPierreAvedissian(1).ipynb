```python
def somme(n):
    somme=0
    for i in range (0,n+1):
        somme=somme+i
    return somme

somme(5)
```

```python
def moyenne(n):
    if n==0:
        print("erreur, sélectionner n supérieur à 0")
    somme=0
    for i in range (1,n+1):
        somme=somme+i
    return somme/n

moyenne(80)
```

```python
def somme_n_rec(n):
    if n==0:
        return 0
    return somme_n_rec(n-1)+n


def Moyenne_rec(n):
    if n==0:
        print ("erreur: choisir n supérieur à 0")
        return None
    return somme_n_rec(n)/n


n=5
print("la moyenne des" ,n, "premiers nombres est:")
Moyenne_rec(n) 
```

```python
def fact(n):
    if n==0:
        return 1
    fact=1
    for i in range (1,n+1):
        fact = fact*i
    return fact
    
n=63
print("Le factoriel de" ,n, "est égal à" ,fact(n),)
```

```python
def fibo_rec(n):
    if n==0:
        return 0
    if n==1:
        return 1
    return fibo_rec(n-1) + fibo_rec(n-2)

fibo_rec(5)
```
